```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("D:\PROYECTOS\SPOTIFY-ANALYSIS\data\dataset.csv")

print(df.head())
print(df.info())
print(df.describe())
print(df.dtypes)
print(df.shape)

```

```python
print("Canciones que duran 0 segundos:")
print(df[df['duration_ms'] == 0])

print("CAnciones con tempo 0:")
print(df[df['tempo'] == 0][['track_name', 'artists', 'track_genre', 'tempo', 'duration_ms']])

print("Observamos cuantas son en total")
print(f"\nCanciones con duration_ms = 0: {(df['duration_ms'] == 0).sum()}")
print(f"Canciones totales con tempo 0: {(df['tempo'] == 0).sum()}")
```

```python
# miramos cual es el genero con ma´s canciones con tempo 0
print(df[df['tempo'] == 0]['track_genre'].value_counts())

# Cómo son sus otras features de media comparado con el resto
print("\nMedia en canciones con tempo 0 ")
print(df[df['tempo'] == 0][['energy', 'valence', 'danceability', 'acousticness']].mean())

print("\nMedia en el resto del dataset")
print(df[df['tempo'] != 0][['energy', 'valence', 'danceability', 'acousticness']].mean())

# miramos géneros totales
print(f"Total de géneros: {df['track_genre'].nunique()}")

print(df['track_genre'].value_counts())
```

```python
# miramos bien los géneros del dataset
print(sorted(df['track_genre'].unique()))

# miramos en guitar y piano
print("Género PIANO")
print(df[df['track_genre'] == 'piano'][ ['track_name', 'artists', 'tempo', 'duration_ms']].head())

print("Género GUITARRA")
print(df[df['track_genre'] == 'guitar'][ ['track_name', 'artists', 'tempo', 'duration_ms']].head(10))
```

```python
# una vez elegidos los géneros, eliminamos los que no nos interesan
generos_a_eliminar =[
    'sleep',
    'ambient', 
    'comedy',
    'children',
    'kids',
    'study',
    'new-age'
]

# primero eliminamos la columna con el índice innecesario
df = df.drop(columns=['Unnamed: 0'])

# eliminarmos los géneros que no nos interesan
df = df[~df['track_genre'].isin(generos_a_eliminar)]

# eliminamos la canción con duration_ms = 0
df = df[df['duration_ms'] > 0]

df = df.dropna()

# Verificamos resultado
print(f"Filas originales: 114000")
print(f"Filas tras limpieza: {len(df)}")
print(f"Géneros restantes: {df['track_genre'].nunique()}")
print(f"Nulos restantes: {df.isnull().sum().sum()}")
```

```python
# guardamos el nuevo dataset limpio
df.to_csv("D:\PROYECTOS\SPOTIFY-ANALYSIS\data\dataset_limpio.csv", index=False)
print("Dataset limpio guardado correctamente.")
```
